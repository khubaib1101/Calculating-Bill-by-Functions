```python

# Hospital Bill 11
def calculate_bill(doctor_type,lab_test,medicine):
  doc_payment=0
  doctor_type = doctor_type.lower()
  if doctor_type == 'general':
    doc_payment = 1500
  elif doctor_type == 'specialist' :
    doc_payment = 3000
  elif doctor_type == 'emergency' :
    doc_payment = 5000
  if lab_test == True:

    lab_payment = 1000
  elif lab_test == False:

      lab_payment = 0
  if medicine == True:

    med_payment = 2000
  elif medicine == False:

      med_payment = 0

  total = doc_payment + lab_payment + med_payment
  print("doctor type:",doctor_type)
  print("lab test: ",lab_test)
  print('medicine:',medicine)
  print("____________________________________")
  print("Doctor Fee",doc_payment)
  print("Lab Fee",lab_payment)
  print("Medicine Fee",med_payment)
  print("Total:",total)
calculate_bill('Specialist',True,True)
```
